```python
query = """
SELECT Region, SUM(Sales) AS Total_sales
FROM sales_data
GROUP BY Region
ORDER BY Total_sales DESC
LIMIT 5
"""
result = pd.read_sql(query, con=engine)
result
```

```python
#cateo

query="""
SELECT Category , SUM(Sales) AS Total_sales 
FROM sales_data
GROUP BY Category
ORDER BY Total_sales DESC LIMIT 5
"""
result = pd.read_sql(query, con=engine)
result
```

```python
# overall monthly sales

query="""
SELECT Order_month , SUM(Sales) AS Total_sales 
FROM sales_data
GROUP BY Order_month
ORDER BY Total_sales DESC LIMIT 5
"""
result = pd.read_sql(query, con=engine)
result
```

```python
# state wise sales 

query="""
SELECT State , SUM(Sales) AS Total_sales 
FROM sales_data
GROUP BY State
ORDER BY Total_sales DESC LIMIT 5
"""
result = pd.read_sql(query, con=engine)
result
```

```python
# Top 5 customers 

query="""
SELECT * FROM (SELECT `Customer Name`, `Customer ID`, SUM(Sales) AS total_purchase_amt,
DENSE_RANK() OVER(ORDER BY SUM(Sales) DESC) AS overall_rank
FROM sales_data
GROUP BY `Customer Name`, `Customer ID`) t
WHERE t.overall_rank <6
ORDER BY  t.overall_rank ,t.`Customer Name` """

result=pd.read_sql(query, con=engine)
result
```

```python
data["Order Date"].dt.year.unique()
```

```python
# yearly and monthly sales 

query="""SELECT YEAR(`Order Date`) AS yr, MONTH(`Order Date`) AS mn, SUM(Sales) AS monthly_sales
FROM sales_data
GROUP BY YEAR(`Order Date`), MONTH(`Order Date`)
ORDER BY yr, mn """

result=pd.read_sql(query, con=engine)
result
```

```python
# Month by Month growth

query= """
SELECT yr,mn,monthly_sales,prev_month_sales,
ROUND(((monthly_sales - prev_month_sales)/prev_month_sales)*100,2) AS growth_percentage
FROM(
SELECT yr,mn,monthly_sales,
LAG(monthly_sales) OVER(ORDER BY yr,mn) AS prev_month_sales
FROM (SELECT YEAR(`Order Date`) AS yr, MONTH(`Order Date`) AS mn, SUM(Sales) AS monthly_sales
FROM sales_data
GROUP BY YEAR(`Order Date`), MONTH(`Order Date`)
ORDER BY yr, mn) AS monthly_data ) AS with_lag 
"""

result = pd.read_sql(query, con=engine)          
result
```
